# Data Visualization in Python

In this notebook we use the `tips` dataset to make different charts with Matplotlib, Seaborn, and Plotly.

## 1. Import the libraries

```python

import matplotlib.pyplot as plt
```

```python

import seaborn as sns
```

```python

import plotly.express as px
```

```python

import pandas as pd
```

## 2. Load the dataset

```python

df = sns.load_dataset("tips")
df.head()
```

## 3. Bar chart
This chart shows the average tip for each day of the week.

```python

avg_tip = df.groupby("day")["tip"].mean()


plt.bar(avg_tip.index, avg_tip.values)
plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.show()
```

## 4. Line chart
This chart shows the total bill for the first 50 records.

```python

first_50 = df.head(50)


plt.plot(first_50.index, first_50["total_bill"])
plt.title("Total Bill for the First 50 Records")
plt.xlabel("Record Number")
plt.ylabel("Total Bill")
plt.show()
```

## 5. Histogram
This chart shows how the total bill amounts are spread out.

```python

plt.hist(df["total_bill"], bins=20)
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

## 6. Scatter plot
This chart shows the relationship between the total bill and the tip.

```python

plt.scatter(df["total_bill"], df["tip"])
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

## 7. Box plot
This chart shows the total bill amounts for each day, and helps us see outliers.

```python

sns.boxplot(x="day", y="total_bill", data=df)
plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

## 8. Pie chart
This chart shows the percentage of smokers and non-smokers.

```python

smoker_counts = df["smoker"].value_counts()


plt.pie(smoker_counts, labels=smoker_counts.index, autopct="%1.1f%%")
plt.title("Smokers and Non-Smokers")
plt.show()
```

## 9. Heatmap
This chart shows how total bill, tip, and party size relate to each other.

```python

correlation = df[["total_bill", "tip", "size"]].corr()


sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

## 10. Interactive Plotly chart
This chart shows the same total bill vs tip data, but you can zoom and hover over the points.

```python

fig = px.scatter(df, x="total_bill", y="tip")
fig.show()
```
